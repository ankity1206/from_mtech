```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Generate the random data
N = 50000
X = np.random.uniform(-1, 1, N)
Y = np.random.uniform(-1, 1, N)
Z = X**2 + Y**2

# 2. Set up the plotting canvas
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Plot 1: Spatial Scatter Plot colored by Z value
scatter = ax1.scatter(X, Y, c=Z, cmap='viridis', s=1, alpha=0.6)
ax1.set_title("Points (X, Y) Color-Coded by Z")
ax1.set_xlabel("X")
ax1.set_ylabel("Y")
ax1.set_aspect('equal')
fig.colorbar(scatter, ax=ax1, label='$Z = X^2 + Y^2$')

# Plot 2: Histogram showing the Probability Density of Z
ax2.hist(Z, bins=100, density=True, color='purple', alpha=0.7, edgecolor='black', linewidth=0.2)
ax2.axvline(1.0, color='red', linestyle='--', label='Truncation Edge (Z=1)')
ax2.set_title("Probability Density Function (PDF) of Z")
ax2.set_xlabel("Z Value")
ax2.set_ylabel("Density")
ax2.legend()

plt.tight_layout()
plt.show()
```
